# Data Cleaning

```python
import pandas as pd
from pathlib import Path

data_path = Path("../data/raw")

csv_files = list(data_path.glob("*.csv"))

dataframes = []

for file in csv_files:
    df_temp = pd.read_csv(file)
    dataframes.append(df_temp)

df = pd.concat(dataframes, ignore_index=True)

df.columns = (
    df.columns
    .str.strip()
    .str.replace('\ufeff', '', regex=False)
)

print("Dataset loaded successfully!")
print(df.shape)
```

```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
import numpy as np

numeric_columns = df.select_dtypes(include=np.number).columns

infinite_values = np.isinf(df[numeric_columns]).sum()

print(infinite_values[infinite_values > 0])
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
import numpy as np

df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
print("Total missing values after replacing infinity:",
      df.isnull().sum().sum())
```

```python
df.dropna(inplace=True)
```

```python
print("Total missing values after cleaning:",
      df.isnull().sum().sum())

print("Dataset shape after removing missing values:",
      df.shape)
```

```python
df.drop_duplicates(inplace=True)
```

```python
print("Number of duplicate rows after cleaning:", df.duplicated().sum())
print("Final cleaned dataset shape:", df.shape)
```

```python
df.drop_duplicates(inplace=True)
```

```python
from pathlib import Path

processed_path = Path("../data/processed")

output_file = processed_path / "cleaned_cicids2017.csv"

df.to_csv(output_file, index=False)

print(f"Cleaned dataset saved to: {output_file}")
```
